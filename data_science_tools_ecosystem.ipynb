## Arithmetic Expression Examples
The next cells show multiplication, addition, and simple unit conversion.


```
# Multiply and add numbers
a = 3
b = 4
c = 5
result = a * b + c  # (3 * 4) + 5 = 17
print("Result:", result)
```

```
# Convert minutes to hours
minutes = 200
hours = minutes / 60
whole_hours = minutes // 60
remaining_minutes = minutes % 60

print("Hours (decimal):", hours)
print(f"{minutes} minutes = {whole_hours} hour(s) and {remaining_minutes} minute(s)")
```
